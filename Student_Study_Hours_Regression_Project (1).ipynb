```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression as SkLinearRegression
from sklearn.linear_model import LogisticRegression as SkLogisticRegression
```

## 1. Linear Regression from scratch

Linear Regression predicts a numerical value. Here, the input is study hours and the target is marks.

```python
class LinearRegression:

    def fit(self, X, y):
        x_mean = np.mean(X)
        y_mean = np.mean(y)

        numerator = np.sum((X - x_mean) * (y - y_mean))
        denominator = np.sum((X - x_mean) ** 2)

        self.m = numerator / denominator
        self.c = y_mean - (self.m * x_mean)

    def predict(self, x):
        return (self.m * x) + self.c
```

## 2. Logistic Regression from scratch

Logistic Regression predicts a class. Here, class `1` means Pass (marks 50 or above), and class `0` means Not Pass.

```python
class LogisticRegression:

    def fit(self, X, y):
        self.m = 0.0
        self.c = 0.0

        learning_rate = 0.01
        epochs = 3000
        n = len(X)

        for i in range(epochs):
            y_pred = self.m * X + self.c
            probability = 1 / (1 + np.exp(-y_pred))
            error = probability - y

            dm = np.sum(error * X) / n
            dc = np.sum(error) / n

            self.m = self.m - learning_rate * dm
            self.c = self.c - learning_rate * dc

    def predict(self, x):
        score = (self.m * x) + self.c
        probability = 1 / (1 + np.exp(-score))
        prediction = 1 if probability >= 0.5 else 0
        return probability, prediction
```

## 3. Create the practice dataset

`study_hours` is the feature (X). `marks` is the numerical target (y). The pass/fail target is created from marks.

```python
df = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "marks":       [18, 25, 32, 42, 51, 60, 68, 78, 88]
})

df["result"] = (df["marks"] >= 50).astype(int)
df
```

## 4. Linear Regression prediction

```python
X = df["study_hours"]
y = df["marks"]

linear_reg = LinearRegression()
linear_reg.fit(X, y)

hours = 5.5
prediction = linear_reg.predict(hours)

print(f"Study hours: {hours}")
print(f"Predicted marks (scratch): {prediction:.2f}")

sk_linear_model = SkLinearRegression()
sk_linear_model.fit(df[["study_hours"]], y)

sk_prediction = sk_linear_model.predict(pd.DataFrame({"study_hours": [hours]}))
print("Scikit-learn prediction:", round(sk_prediction[0], 2))
```

## 5. Logistic Regression prediction

```python
y_class = df["result"]

logistic_reg = LogisticRegression()
logistic_reg.fit(X, y_class)

hours = 5.5
probability, prediction = logistic_reg.predict(hours)

print(f"Study hours: {hours}")
print(f"Pass probability (scratch): {probability:.2f}")
print("Predicted result:", "Pass" if prediction == 1 else "Not Pass")

sk_logistic_model = SkLogisticRegression()
sk_logistic_model.fit(df[["study_hours"]], y_class)

test = pd.DataFrame({"study_hours": [hours]})
sk_probability = sk_logistic_model.predict_proba(test)[0][1]
sk_class = sk_logistic_model.predict(test)[0]

print("\nScikit-learn probability:", round(sk_probability, 2))
print("Scikit-learn result:", "Pass" if sk_class == 1 else "Not Pass")
```
